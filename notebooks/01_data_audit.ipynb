```python

# Import the required libraries
import pandas as pd
import numpy as np
from pathlib import Path

# Define the location of our raw dataset
DATA_PATH = Path("../data/raw")

# Load the customer and order datasets
customers = pd.read_csv(DATA_PATH / "customers.csv")
orders = pd.read_csv(DATA_PATH / "orders.csv")

# Display successful loading information
print("Datasets loaded successfully!")

print("\nCustomers Dataset:")
print("Rows:", customers.shape[0])
print("Columns:", customers.shape[1])

print("\nOrders Dataset:")
print("Rows:", orders.shape[0])
print("Columns:", orders.shape[1])
```

```python

print("MISSING VALUES IN CUSTOMERS")
print(customers.isnull().sum())

print("\nMISSING VALUES IN ORDERS")
print(orders.isnull().sum())
```

```python

# Calculate missing value percentages

print("CUSTOMERS - MISSING VALUE PERCENTAGE")

customer_missing = (
    customers.isnull().sum() / len(customers)
) * 100

print(customer_missing[customer_missing > 0])


print("\nORDERS - MISSING VALUE PERCENTAGE")

orders_missing = (
    orders.isnull().sum() / len(orders)
) * 100

print(orders_missing[orders_missing > 0])
```

```python

# Check duplicate records

print("CUSTOMERS DUPLICATES:",
      customers.duplicated().sum())

print("ORDERS DUPLICATES:",
      orders.duplicated().sum())
```

```python

# Check completely duplicated records

print("CUSTOMERS - Duplicate Rows:",
      customers.duplicated().sum())

print("ORDERS - Duplicate Rows:",
      orders.duplicated().sum())
```

```python

# Check duplicate primary keys

print("Duplicate Customer IDs:",
      customers["customer_id"].duplicated().sum())

print("Duplicate Order IDs:",
      orders["order_id"].duplicated().sum())
```

```python

# Check missing primary keys

print("Missing Customer IDs:",
      customers["customer_id"].isnull().sum())

print("Missing Order IDs:",
      orders["order_id"].isnull().sum())
```

```python

# Check whether every order belongs to an existing customer

invalid_customer_ids = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

print("Orders with invalid customer IDs:",
      len(invalid_customer_ids))
```

```python

# Count actual orders placed by each customer

actual_order_counts = orders.groupby(
    "customer_id"
).size()

print("Total customers who placed orders:",
      actual_order_counts.count())

print("\nAverage orders per purchasing customer:",
      actual_order_counts.mean())

print("\nMaximum orders placed by one customer:",
      actual_order_counts.max())
```

```python

# Compare recorded and actual order counts

customer_order_check = customers[
    ["customer_id", "total_orders"]
].copy()

customer_order_check["actual_orders"] = (
    customer_order_check["customer_id"]
    .map(actual_order_counts)
    .fillna(0)
    .astype(int)
)

customer_order_check["difference"] = (
    customer_order_check["total_orders"]
    - customer_order_check["actual_orders"]
)

print("Customers with mismatched order counts:",
      (customer_order_check["difference"] != 0).sum())

customer_order_check.head(10)
```

```python

# Calculate expected final amount

orders["calculated_amount"] = (
    orders["subtotal"]
    + orders["shipping_fee"]
    + orders["tax_amount"]
).round(2)

# Compare with recorded final amount

orders["amount_difference"] = (
    orders["final_amount"]
    - orders["calculated_amount"]
).round(2)

# Count mismatches

mismatches = orders[
    orders["amount_difference"] != 0
]

print("Total orders:", len(orders))
print("Orders with incorrect final amount:", len(mismatches))
```

```python

# Load order items dataset
order_items = pd.read_csv(DATA_PATH / "order_items.csv")

# Calculate expected profit
order_items["calculated_profit"] = (
    order_items["item_revenue"]
    - order_items["item_cost"]
).round(2)

# Identify mismatches
profit_mismatches = order_items[
    order_items["profit"].round(2)
    != order_items["calculated_profit"]
]

print("Total order items:", len(order_items))
print("Incorrect profit records:", len(profit_mismatches))
```

```python
# Load the remaining datasets
products = pd.read_csv(DATA_PATH / "products.csv")
payments = pd.read_csv(DATA_PATH / "payments.csv")
shipments = pd.read_csv(DATA_PATH / "shipments.csv")
returns = pd.read_csv(DATA_PATH / "returns.csv")
customer_reviews = pd.read_csv(DATA_PATH / "customer_reviews.csv")
marketing_campaigns = pd.read_csv(DATA_PATH / "marketing_campaigns.csv")

# Display dataset sizes
datasets = {
    "Products": products,
    "Payments": payments,
    "Shipments": shipments,
    "Returns": returns,
    "Customer Reviews": customer_reviews,
    "Marketing Campaigns": marketing_campaigns
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
all_datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Products": products,
    "Payments": payments,
    "Shipments": shipments,
    "Returns": returns,
    "Customer Reviews": customer_reviews,
    "Marketing Campaigns": marketing_campaigns
}

for name, df in all_datasets.items():
    total_missing = df.isnull().sum().sum()
    columns_with_missing = (df.isnull().sum() > 0).sum()

    print(f"\n{name}")
    print("Total missing values:", total_missing)
    print("Columns containing missing values:", columns_with_missing)
```

```python
for name, df in all_datasets.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if not missing.empty:
        print(f"\n{'='*45}")
        print(name)
        print("="*45)

        for column, count in missing.items():
            percentage = (count / len(df)) * 100
            print(f"{column}: {count} missing ({percentage:.2f}%)")
```

```python
total_cost = marketing_campaigns["campaign_cost"].sum()
total_impressions = marketing_campaigns["impressions"].sum()
total_clicks = marketing_campaigns["clicks"].sum()
total_conversions = marketing_campaigns["conversions"].sum()

ctr = (total_clicks / total_impressions) * 100
conversion_rate = (total_conversions / total_clicks) * 100
cost_per_conversion = total_cost / total_conversions

print("MARKETING PERFORMANCE KPIs")

print(f"Click-Through Rate (CTR): {ctr:.2f}%")
print(f"Conversion Rate: {conversion_rate:.2f}%")
print(f"Cost per Conversion: ₹{cost_per_conversion:.2f}")
```

```python
print("DATE RANGE ANALYSIS")

date_columns = {
    "Customers": (customers, "customer_signup_date"),
    "Orders": (orders, "order_date"),
    "Products": (products, "product_launch_date"),
    "Shipments": (shipments, "dispatch_date"),
    "Returns": (returns, "return_date"),
    "Customer Reviews": (customer_reviews, "review_date"),
    "Marketing Campaigns": (marketing_campaigns, "campaign_start_date")
}

for name, (df, column) in date_columns.items():
    dates = pd.to_datetime(df[column], errors="coerce")

    print(f"\n{name}")
    print("Earliest date:", dates.min())
    print("Latest date:", dates.max())
    print("Invalid dates:", dates.isnull().sum())
```

```python
date_audit = []

for name, (df, column) in date_columns.items():
    dates = pd.to_datetime(df[column], errors="coerce")

    date_audit.append({
        "Dataset": name,
        "Date Column": column,
        "Earliest Date": dates.min(),
        "Latest Date": dates.max(),
        "Invalid Dates": dates.isnull().sum()
    })

date_audit_df = pd.DataFrame(date_audit)

display(date_audit_df)
```
